* Division by subtraction: An alien approaches you and wants to learn to compute
N divided by m but they only know addition and subtraction. How do you teach them
to divide N by m? Write an algorithm that uses only addition/subtraction that takes
in two positive integers N, m and outputs the q, r ∈ Z such that
N = q · m + r.

```python
def naive_div(N,m):
    r = N
    q = 0
    while r >= m:
        r = r - m
        q = q + 1
    
    return q,r

#Testing function
N, m = 100, 34
q, r = naive_div(N,m)
print(N,"divided by",m,"=", q, "remainder", r)
print("N = q * m + r is", (q * m + r))
```

* Euclid’s algorithm: Write a python function that takes in two integers a, b and
outputs GCD(a, b) using Euclid’s algorithm.

```python
#Euclid's alg which returns gcd of a and b
def Euclid(a, b):
    if a == 0:
        return b
 
    return Euclid(b % a, a)

# Example usage:
a, b = 352, 79
gcd = Euclid(a, b)
print("GCD of", a, "and", b, "is:", gcd)
a, b = 362, 94
gcd = Euclid(a, b)
print("GCD of", a, "and", b, "is:", gcd)
```
