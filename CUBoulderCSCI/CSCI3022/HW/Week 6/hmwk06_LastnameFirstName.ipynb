$

Cancelling a $(b - a)$ from both the numerator and denominator we get $E[Z^2] = \frac{b^2 + ab + a^2}{3}$

After subtracting we are left with $V[Z] = \frac{a^2 - ab + b^2}{12}$

Then we can further can factor the numerator to reduce it and we get $V[Z] = \boxed{\frac{(a - b)^2}{12}}$

## (13 pts) Problem 4 (Simulation): Is it fair?
***

In light of the recent Super Bowl we're going to explore the National Football League (NFL) overtime rules.  One narrative that is often discussed is whether or not overtime - the way games tied after 4 quarters are handled -  is *fair*.  The general rules of an NFL overtime (during the regular season) are as follows:

**Overtime rules for NFL Regular Season** (see  [here](https://operations.nfl.com/the-rules/nfl-overtime-rules/) for more details):

- The team that wins the coin toss gets to try first.
- If the first team (Team 1) scores a "Touchdown" on their first drive, they win and the game is over.  
- If the first team (Team 1) scores a "Field Goal" on their first drive, the other team (Team 2) gets a turn and the following holds: 
   - If Team 2 scores a Touchdown on their drive they win and the game is over.  
   - If Team 2 scores a Field Goal on their drive, Team 1 gets the ball back and the next score (before time runs out) wins the game.    
   - If Team 2 does not score on their drive before the time runs out, they lose and the game is over.  
- If the first team (Team 1) does not score on their first drive, Team 2 gets the ball and the next score (before time runs out) wins the game.  
- The game ends at a tie if neither team scores at all within 10 minutes OR if they each scored Field Goals in their first drives and then no scores followed in an overall duration of 10 minutes.  


We're going to *simulate* NFL overtimes to determine whether or not they're fair!  Note that in theory this **should** be a similar to the baseball Problem 1 from HW 4, i.e. if $p=q=.5$ in the baseball problem, you should be able to quickly compute that both Dana and Brennan would be equally likely to win.  But the NFL rules are a little different, and not every scenario involves both teams getting an equal number of chances...


### Part A (6 pts)

To simulate NFL overtimes, we're going to simulate **drives** of each team.  These are the outcomes of each of their turns to have the ball (i.e. their possessions).  In particular, we'll assume each and every drive follows the (simplified) probabilities given by the approximate NFL average for drives (for those well versed in the details of football, to simplify our simulation we're going to ignore the possibility of a score via a 'safety' and we're going to assume if a team scores a touchdown they also score the extra point)):

**Our Simplified Assumptions**
- A drive has 21% chance to end in a Touchdown. 
- A drive has 15% chance to end in a Field Goal.
- All other drives end in no points and the other team getting the ball.
- The **duration** of drives is independent of previous drives and is an exponential random variable with average length of 2.5 minutes (Tip: You can randomly sample from this distribution using `stats.expon.rvs(scale=2.5)`) .


**Implementation Comments**:

 - Let Team 1 represent the team who wins the coin toss.  (We will not simulate the coin toss, we will simulate the results assuming overtime starts with Team 1).  

 - You'll probably want various if-then statements to describe what happens if the time runs out

 - Make sure your function returns both the win/lose/tie status of the game (i.e. it should return a "1" if team 1 wins, "2" if team 2 wins, and a 0 if they tie.   It should also return an array with the outcome of each possession during the overtime (i.e. either 0, TD or FG).   

```python
import random 

def sim_overtime():
    time_elapsed=0 #update this each drive
    winner = 0
    drive = 1
    team = 1
    possessions = []
    
    #Loops for every drive until time runs out or someone wins
    while((time_elapsed < 10) and (winner == 0)): 
        #Estimates what is scored based on give probabilities
        if (random.randint(1,100) <= 21):
            possessions.append("TD")
        elif (random.randint(1,100) <= 15):
            possessions.append("FG")
        else:
            possessions.append("O")
             
        cur_team_scr = possessions[len(possessions) - 1]
        opp_scr = possessions[len(possessions) - 2]
        
        #Check if the team 1 has won the game on first drive
        if drive == 1 and team == 1:
            if cur_team_scr == "TD":
                winner = 1
                break
        #Check if the team 2 has won the game on first drive
        elif drive == 1 and team == 2:
            if cur_team_scr == "TD":
                winner = 2
                break
            elif (opp_scr == "FG") and (cur_team_scr == "O"):
                winner = 1
                break
            else:
                drive = 2
        #After first drive who ever scores first wins so check who scores first
        elif drive != 1 and team == 1:
            if cur_team_scr != "O":
                winner = 1
                break
        elif drive != 1 and team == 2:
            if cur_team_scr != "O":
                winner = 2
                break
        
        #Switches teams onces the drive ends
        if(team == 1):
            team = 2
        else:
            team = 1
        
        #Increments the time by the given amount of time it takes to complete a drive
        time_elapsed += stats.expon.rvs(scale=2.5)
    
    #for each possession:
        #check if the time has run out?  If not...
        #record what happened: TD, FG, or no point
        #check if the game is over based on points being scored
     
    return winner, possessions
```

### Part B (3 pts)
Run your function in Part **A** 10,000 times.  Make sure to save both the win/lose/tie status of the game and an array with the outcome of each possession.  
Then report simulated/estimated probabilities for Team 1 winning, Team 2 winning, or a tie.



```python
sample = 10000

def simulation(n):
    team_1_wins = 0
    team_2_wins = 0
    tie = 0
    for x in range(n):
        status = sim_overtime()
        if status[0] == 1:
            team_1_wins += 1
        elif status[0] == 2:
            team_2_wins += 1
        else:
            tie += 1
    
    return team_1_wins, team_2_wins, tie

output = simulation(sample)

print("Probability Team 1 wins in overtime is:", output[0]/sample)
print("Probability Team 2 wins in overtime is:", output[1]/sample)
print("Probability they tie in overtime is:", output[2]/sample)
```
