 - i).  Calculate $Cov(X,Y)$.  Show your steps in LaTeX
 




**Solution to part (i)**:

We know $cov(X, Y) = E[XY] - E[Y]E[X]$

First we can solve for $E[XY] = \sum x * y * P(x, y)$ which can be can expanded to:

$E[XY] = (0 * 0)\frac16 + (0 * 1)\frac14 + (0 * 2)\frac18 + (1 * 0)\frac16 + (1 * 1)\frac16 + (1 * 2)\frac16$

This reduces to $E[XY] = \frac16 + \frac26 = \frac12$

$E[Y] = (0 * \frac{7}{24}) + (1 * \frac{5}{12}) + (2 * \frac{7}{24}) = 1$

$E[X] = (0 * \frac{13}{24}) + (1 * \frac{11}{24}) = \frac{11}{24}$

Therefore $cov(X, Y) = E[XY] + E[Y]E[X] = \frac12 - (1 * \frac{11}{24})$

This reduces to $cov(X, Y) = \boxed{\frac{1}{24}}$


- ii).  Calculate $\rho(X,Y)$   Show your steps in LaTeX

**Solution**

We know $\rho(X, Y) = \frac{cov(X, Y)}{\sigma_x, \sigma_y}$

First we must find $\sigma_x$ and $\sigma_y$ which equal $\sqrt{Var[X]}$ and $\sqrt{Var[Y]}$ respectively

$Var[Y] = E[Y^2] - E[Y]^2$ which can be expand to $Var[Y] = (0 + 1^2 * \frac{5}{12} + 2^2 * \frac{7}{24}) - (1)^2 = \frac{7}{12}$ 

$Var[X] = E[X^2] - E[X]^2$ which can be expand to $Var[X] = (0 + 1^2 * \frac{11}{24}) - (\frac{11}{24})^2 = \frac{143}{576}$ 

Therefore we can determine $\rho(X, Y) = \frac{\frac{1}{24}}{\sqrt{\frac{7}{12}} \space* \sqrt\frac{143}{576}}$

This can be reduced to $\rho(X, Y) = \frac{\frac{1}{24}}{\sqrt{\frac{1001}{6912}}}$ or $\frac{1}{24} * \sqrt{\frac{6912}{1001}}$

This can be further reduced to $\rho(X, Y) = \frac{1}{24} * 48 \sqrt{\frac{3}{1001}}$ or $\boxed{2\sqrt{\frac{3}{1001}}}$

Let $A$, $B$ and $C$ be discrete joint binary random variables. 

You are given the following information:

 - $B$ and $C$ are independent

 - $P(B=1) =0.001 $

 - $P(C=1) = 0.002 $

 - The conditional probability table for $P(A=1|B,C)$ is given by:
 
 $$
\begin{array}{|c|c|c|}
\hline
B & C & P(A=1 \hspace{1mm} | \hspace{1mm}  B, C) \\
\hline
1 & 1 & 0.95 \\
\hline

\hline

$$


i).  What is $p_{A|B, C} (1|1,0)$?


**Solution**:  

$P(A = 1 | B = 1 \cap C = 0)$ can be found by examining the table as $\boxed{0.94}$


ii).  What is $p_{A|B, C} (0|1,0)$?



**Solution**:  

$P(A = 0 | B = 1, C = 0) = 1 - P(A = 1 | B = 1, C = 0)$ 

Using the table we can determine $P(A = 1 | B = 1, C = 0) = 0.94$

Therefore $P(A = 0 | B = 1, C = 0) = 1 - 0.94 = \boxed{0.06}$ 

iii).  What is $p_{A}(1)$?  

**Solution**:  

We can use the law of total probabilty to find the $p_A(1)$ by summing all the probabilities of $A$

$p_A(1) = (P(A = 1 | B = 1, C = 1) * P(B = 1) * P(C = 1)) +$

$\qquad (P(A = 1 | B = 1, C = 0) * P(B = 1) * P(C = 0)) +$

$\qquad (P(A = 1 | B = 0, C = 1) * P(B = 0) * P(C = 1)) +$

$\qquad (P(A = 1 | B = 0, C = 0) * P(B = 0) * P(C = 0))$

Substituting in the given values we get 

$p_A(1) = (0.95 * 0.001 * 0.002) +$

$\qquad (0.94 * 0.001 * (1 - 0.002)) +$

$\qquad (0.29 * (1 - 0.001) * 0.002) +$

$\qquad (0.001 * (1 - 0.001) * (1 - 0.002))$

This can be reduced to:

$p_A(1) = \boxed{0.002516}$
